```python
import os
import pandas as pd

# Local paths
DATA_ROOT    = r'C:\Users\GHANSHYAM\Desktop\voice-clone-detector\data\raw\LA\LA'
PROTOCOL_DIR = f'{DATA_ROOT}\\ASVspoof2019_LA_cm_protocols'
TRAIN_AUDIO  = f'{DATA_ROOT}\\ASVspoof2019_LA_train\\flac'

# verify
print(os.listdir(PROTOCOL_DIR))
print(os.listdir(TRAIN_AUDIO)[:5])
```

```python
train_df = pd.read_csv(
    f'{PROTOCOL_DIR}\\ASVspoof2019.LA.cm.train.trn.txt',
    sep=' ',
    header=None,
    names=['speaker_id', 'file_id', 'env', 'attack_id', 'label']
)

dev_df = pd.read_csv(
    f'{PROTOCOL_DIR}\\ASVspoof2019.LA.cm.dev.trl.txt',
    sep=' ',
    header=None,
    names=['speaker_id', 'file_id', 'env', 'attack_id', 'label']
)

eval_df = pd.read_csv(
    f'{PROTOCOL_DIR}\\ASVspoof2019.LA.cm.eval.trl.txt',
    sep=' ',
    header=None,
    names=['speaker_id', 'file_id', 'env', 'attack_id', 'label']
)

print("=== TRAIN ===")
print(train_df['label'].value_counts())
print(f"Shape: {train_df.shape}")

print("\n=== DEV ===")
print(dev_df['label'].value_counts())
print(f"Shape: {dev_df.shape}")

print("\n=== EVAL ===")
print(eval_df['label'].value_counts())
print(f"Shape: {eval_df.shape}")

print("\n=== ATTACK TYPES ===")
print(train_df['attack_id'].value_counts())
```

```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

# ek real aur ek fake file
real_file = train_df[train_df['label'] == 'bonafide'].iloc[0]['file_id']
fake_file = train_df[train_df['label'] == 'spoof'].iloc[0]['file_id']

y_real, sr = librosa.load(f'{TRAIN_AUDIO}\\{real_file}.flac', sr=16000)
y_fake, sr = librosa.load(f'{TRAIN_AUDIO}\\{fake_file}.flac', sr=16000)

fig, axes = plt.subplots(2, 2, figsize=(14, 8))

librosa.display.waveshow(y_real, sr=sr, ax=axes[0][0])
axes[0][0].set_title('Real voice — waveform')

librosa.display.waveshow(y_fake, sr=sr, ax=axes[0][1])
axes[0][1].set_title('Fake voice — waveform')

mel_real = librosa.power_to_db(
    librosa.feature.melspectrogram(y=y_real, sr=sr, n_mels=128), ref=np.max)
mel_fake = librosa.power_to_db(
    librosa.feature.melspectrogram(y=y_fake, sr=sr, n_mels=128), ref=np.max)

librosa.display.specshow(mel_real, sr=sr, ax=axes[1][0], x_axis='time', y_axis='mel')
axes[1][0].set_title('Real voice — mel spectrogram')

librosa.display.specshow(mel_fake, sr=sr, ax=axes[1][1], x_axis='time', y_axis='mel')
axes[1][1].set_title('Fake voice — mel spectrogram')

plt.tight_layout()
plt.savefig(r'C:\Users\GHANSHYAM\Desktop\voice-clone-detector\results\figures\real_vs_fake.png', dpi=150)
plt.show()

print(f"Real duration: {len(y_real)/sr:.2f}s")
print(f"Fake duration: {len(y_fake)/sr:.2f}s")
```
